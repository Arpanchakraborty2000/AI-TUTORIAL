```python
import nltk
from nltk.tokenize import word_tokenize
from nltk import pos_tag, ne_chunk

text = "Barack Obama was born in Hawaii"

words = word_tokenize(text)
tags = pos_tag(words)

ner = ne_chunk(tags)

print(ner)
```

```python
ner.draw()
```
